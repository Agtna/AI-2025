```python
import numpy as np
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.mean(score)
print(x)
```

```python
import numpy as np
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.median(score)
print(x)
```

```python
from scipy import stats
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = stats.mode(score, keepdims=True)
print(x)
```

```python
import numpy as np
data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]

minimum = np.min(data)
maximum = np.max(data)
x = maximum - minimum
print(x)
```

```python
import numpy as np
data =[90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.percentile(data, 90)
print(x)
```

```python
import numpy as np
data = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]

Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1
print(Q1, Q3, IQR)
```

```python
import numpy as np
data = [86, 87, 88, 86, 87, 85, 86]

mean = np.mean(data)
var = np.var(data)
std = np.std(data)
print(mean, var, std)
```

```python
from scipy import stats

data = [1, 2, 3, 4, 5, 5, 6, 2, 3, 4, 5, 6, 6, 4, 3, 2, 1, 4, 3, 5, 6, 6, 5, 4, 5]
x = stats.mode(data, keepdims=True)
print(f"Modus dari data adalah: {x.mode[0]} (muncul sebanyak {x.count[0]} kali)")
```

2. Persentil

```python
import numpy as np

data = [9, 10, 11, 6, 8, 7, 7, 5, 4, 5]
x = np.percentile(data, 25)
y = np.percentile(data, 75)
print(f"Persentil ke-25 dari data adalah: {x}")
print(f"Persentil ke-75 dari data adalah: {y}")
```
