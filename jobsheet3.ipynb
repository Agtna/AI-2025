3 b.Identifikasi nilai first quartile (Q1) dan third quartile (Q3), dan IQR

```python
import numpy as np

data = [10, 50, 50, 50, 50, 51, 52, 53, 53, 54, 57, 59, 59, 60, 61, 62, 62, 63, 63, 63, 65, 65, 65, 65, 99]

Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1
print(f"Q1 : ", Q1)
print(f"Q3 : ", Q3)
print(f"IQR : ", IQR)
```

```python
import numpy as np

Q1 = 51.5
Q3 = 63.0
IQR = 11.5

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f" Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")
```
