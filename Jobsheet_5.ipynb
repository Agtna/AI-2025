## Data Understanding
- Merge tables
---





```python
df_merge1 = pd.merge(df_transactions, df_customers, on='CustomerID', how='left')
df_merge2 = pd.merge(df_merge1, df_gasstations, on='GasStationID', how='left')
df_merged = pd.merge(df_merge2, df_products, on='ProductID', how='left')

df_merged.head()
```

## Soal-2
Buatlah list Top 5 Gas stations yang memiliki nilai transaksi paling banyak
---

```python
df_gasstations = df_merged.groupby('GasStationID')
df_top5_station = df_gasstations[['Amount', 'Price']].sum()
df_top5_station.rename(columns={'Price':'Gas Sales'}, inplace=True)
df_top5_station.sort_values(by='Gas Sales', ascending=False).head()
```

## Soal-3
Buatlah list Top 5 Jenis produk yang memiliki nilai transaksi paling banyak
---

```python
df_products = df_merged.groupby('Description')
df_top5_product = df_products[['Amount', 'Price']].sum()
df_top5_product.rename(columns={'Price':'Sales'}, inplace=True)
df_top5_product.sort_values(by='Sales', ascending=False).head()
```

## Soal-4
Buatlah deskripsi statistik untuk masing masing hari (23,24,25,26) pada dataset transaction_1k
---

```python
df = df_transactions.copy()
df['Date'] = pd.to_datetime(df['Date'])
df['Date'] = df['Date'].dt.day
filtered_df = df[df['Date'].isin([23, 24, 25, 26])]
daily_stats= filtered_df.groupby('Date')[['Amount', 'Price']].describe()
daily_stats
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Hour'] = pd.to_datetime(df['Time'], format='%H:%M:%S').dt.hour
df_grouped = df.groupby(['Date', 'Hour']).agg({
    'Amount': 'sum', 'Price': 'sum'}).rename(columns={'Price': 'Gas Sales'})
best_times = df_grouped.reset_index().sort_values(['Date', 'Amount'], ascending=[True, False])
best_per_day = best_times.groupby('Date').first().reset_index()
best_per_day.set_index('Date', inplace=True)
best_per_day
```
