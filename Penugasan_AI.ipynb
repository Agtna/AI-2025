```python
import pandas as pd

data_pasien = pd.read_csv('data_pasien.csv')
```

```python
data_pasien.info()
```

```python
data_pasien.info()

data_pasien.tail()
```

```python
data_pasien.isnull().sum()
```

```python
data_pasien['Merokok'] = data_pasien['Merokok'].map({'Ya': 1, 'Tidak' : 0})

data_pasien['Olahraga'] = data_pasien['Olahraga'].map({'Ya': 1, 'Tidak': 0})

data_pasien['Penyakit'] = data_pasien['Penyakit'].map({'Ya' : 1, 'Tidak' : 0})
```

```python
data_pasien.tail()
```

```python
data_pasien.drop('ID',axis=1, inplace=True)
```

```python
data_pasien.tail()
```

```python
x = data_pasien[['Usia', 'IMT','TekananDarah', 'Merokok', 'Olahraga']]
y = data_pasien['Penyakit']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.1, random_state=99)
```

```python
from sklearn.preprocessing import MinMaxScaler
sc = MinMaxScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_model = DecisionTreeClassifier()

tree_model.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = tree_model.predict(x_test)

acc_score = round(accuracy_score(y_pred, y_test), 3)

print('Accuracy: ', acc_score)
```

```python
from sklearn. tree import export_graphviz
export_graphviz(
    tree_model,
    out_file = "tree_model.dot",
    feature_names = ['Usia', 'IMT', 'TekananDarah', 'Merokok', 'Olahraga'],
    class_names = ['Tidak', 'Ya'],
    rounded= True,
    filled =True
)
```
